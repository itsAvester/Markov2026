```python
import sympy as sp
import numpy as np
import math
#Problem 2b
av=.04
bv=.16
K=.1
p1=K*math.exp(av*1)
p2=K*math.exp(av*2)
p3=K*math.exp(av*3)
p4=K*math.exp(av*4)
q2=K*math.exp(bv*1)
q3=K*math.exp(bv*2)
q4=K*math.exp(bv*3)
q5=K*math.exp(bv*4)
P=np.array([[1-p1,p1,0,0,0],[q2,1-p2-q2,p2,0,0],[0,q3,1-p3-q3,p3,0],[0,0,q4,1-p4-q4,p4],[0,0,0,q5,1-q5]])
eigvals,eigvecs=np.linalg.eig(P.T)
pi=np.real(eigvecs[:,0])
pi=pi/np.sum(pi)
print(pi)

#Problem 2c
#Start at 1
import matplotlib.pyplot as plt

node = 1
count = np.zeros(5)
n=10000
for i in range(n):
    count[node-1] += 1
    random = np.random.rand()
    if node==1:
        if random<p1:
            node=2
        else:
            node=1
    elif node==2:
        if random<q2:
            node=1
        elif random<(q2+(1-p2-q2)):
            node=2
        else:
            node=3
    elif node==3:
        if random<q3:
            node=2
        elif random<(q3+(1-p3-q3)):
            node=3
        else:
            node=4
    elif node==4:
        if random<q4:
            node = 3
        elif random<(q4+(1-p4-q4)):
            node=4
        else:
            node=5
    elif node==5:
        if random<q5:
            node=4
        else:
            node=5

fractions = count / n

plt.bar([1, 2, 3, 4, 5],fractions)
plt.show()

#Problem 2d
#theory
pi1=1
pi2=1*(p1/q2)
pi3=pi2*(p2/q3)
pi4=pi3*(p3/q4)
pi5=pi4*(p4/q5)

theory = np.array([pi1,pi2,pi3,pi4,pi5])

#all together
plt.bar([1, 2, 3, 4, 5],pi)
plt.show()
plt.bar([1, 2, 3, 4, 5],theory)
plt.show()
plt.bar([1, 2, 3, 4, 5],fractions)
plt.show()

#problem 3b
a = sp.symbols('a', real=True)

A = sp.Matrix([[1-a, a, 0],[a, 0, 1-a],[0, 1-a, a]])

print(A.eigenvects)
print(A.eigenvals)


#Problem 3c (UNFINISHED)
a=.99
A = sp.Matrix([[1-a, a, 0],[a, 0, 1-a],[0, 1-a, a]])

print(A.eigenvects())

q0=np.array([1,0,0])
print(A.eigenvects)
print(A.eigenvals)
ie = A.eigenvals()
iv = A.eigenvects()
```
