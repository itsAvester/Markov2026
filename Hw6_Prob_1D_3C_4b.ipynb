```python
import numpy as np
import matplotlib.pyplot as pt

P=np.array([[0,1,0,0,0],
           [1/3,0,2/3,0,0],
           [0,1/2,0,1/2,0],
           [0,0,2/3,0,1/3],
           [0,0,0,1,0]])
#q0
q=np.array([0,0,0,1,0])
#stationary dist
pi=np.array([1/12,1/4,1/3,1/4,1/12])

P50=np.linalg.matrix_power(P,50)
#q50 just equal P50*q
q50=q@P50

s=np.array([0,1,2,3,4])

pt.plot(s,q50,pi)
```

```python
import numpy as np
a=.49

#extinction count
ext=0
av=1000
gen=200
#10^3 avalanches
for i in range(av):
    xn=1
    #200 generations
    for i in range(gen):
        zn=np.random.binomial(xn,1-a)
        xn=2*zn
        if xn==0:
            ext+=1
            break

frac=ext/av
print(frac)

```
